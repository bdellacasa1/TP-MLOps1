```python
outliers_summary = []

for col in ['Arrival Delay in Minutes', 'Departure Delay in Minutes', 'Flight Distance']:
    Q1 = train_df[col].quantile(0.25)
    Q3 = train_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    n_outliers = (
        (train_df[col] < lower) |
        (train_df[col] > upper)
    ).sum()

    outliers_summary.append([
        col,
        n_outliers,
        round(100 * n_outliers / len(train_df), 2)
    ])

outliers_df = pd.DataFrame(
    outliers_summary,
    columns=["Variable", "Cantidad_Outliers", "Porcentaje"]
)

outliers_df.sort_values(
    by="Porcentaje",
    ascending=False
)
```

Las etapas iniciales de análisis exploratorio y preparación de la estructura de datos se desarrollaron bajo los siguientes criterios metodológicos:

- **Tratamiento de valores faltantes:** Se detectó un 0.3% de datos nulos en la variable Arrival Delay in Minutes. Dado que la distribución de los tiempos de retraso presenta una fuerte asimetría positiva (sesgo a la derecha), se optará por una imputación basada en la mediana para evitar la distorsión que causarían los valores atípicos en la media muestral.

- **Limpieza e ingeniería de características:** Se eliminaron las columnas de control e identificadores (id y Unnamed: 0). Al carecer de capacidad de generalización, mantener estos índices no sumaría información que ayude a los modelos.

- **Codificación de variables (Encoding):**

  - *Variable objetivo*: Se codificó la variable satisfaction para transformarla en una estructura binaria ($0$ y $1$), formato requerido por los clasificadores supervisados.

  - *Variables predictoras*: Las variables categóricas nominales (Gender, Customer Type, etc.) se procesaron mediante One-Hot Encoding. Esto evita introducir un orden jerárquico artificial que sesgue a los estimadores basados en distancias o linealidad.

### **Modelo Base (Baseline)**
Para evaluar si nuestros modelos complejos realmente aportan valor, establecemos un modelo base simple que utiliza una heurística (predecir siempre la clase más frecuente).

```python
#Features para modelo base
X_train_mb = train_df[num_base_cols + score_cols + cat_cols]
X_test_mb = test_df[num_base_cols + score_cols + cat_cols]
y_train_mb =  train_df[target_col]
y_test_mb = test_df[target_col]
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

# El DummyClassifier con strategy='most_frequent' es el estándar para
# establecer la base en clasificación (similar al promedio en regresión).
dummy_clf = DummyClassifier(strategy="most_frequent")
dummy_clf.fit(X_train_mb, y_train_mb)

dummy_acc = accuracy_score(y_test_mb, dummy_clf.predict(X_test_mb))

print(f"--- MODELO BASE ---")
print(f"Accuracy del Modelo Base (Moda): {dummy_acc:.4f}")
print(f"\nEste resultado indica que cualquier modelo útil debe superar el {dummy_acc:.2%} de precisión.")
```

```python
from sklearn.metrics import precision_recall_fscore_support, confusion_matrix
import numpy as np

def get_detailed_metrics(model, X, y_true, name):
    y_pred = model.predict(X)
    acc = accuracy_score(y_true, y_pred)
    # Obtener métricas por clase (0: neutral/dissatisfied, 1: satisfied)
    precision, recall, f1, _ = precision_recall_fscore_support(y_true, y_pred, average='binary')

    # Matriz de confusión para tasas de error
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()

    # Tasa de Falsos Positivos (FPR): Predijimos satisfecho pero no lo estaba
    fpr = fp / (fp + tn)
    # Tasa de Falsos Negativos (FNR): Predijimos insatisfecho pero estaba satisfecho
    fnr = fn / (fn + tp)

    print(f"--- Métricas Detalladas: {name} ---")
    print(f"Accuracy: {acc:.4%}")
    print(f"Precision: {precision:.4%}")
    print(f"Recall: {recall:.4%}")
    print(f"F1-Score: {f1:.4%}")
    print(f"Tasa de Falsos Positivos (FPR): {fpr:.4%}")
    print(f"Tasa de Falsos Negativos (FNR): {fnr:.4%}")
    print(f"Falsos Positivos Absolutos: {fp}")
    print(f"Falsos Negativos Absolutos: {fn}")
    print("-" * 40)

```

```python
get_detailed_metrics(dummy_clf, X_test_mb, y_test_mb, "Modelo Base")
```

### Modelos
Hacemos una primer version teniendo en cuenta la calificacion de los usuarios a los distintos servicios. Como la calificacion es con la encuesta, esto introduce potencial leakage

```python
#Features con scores
features_with_scores = cat_cols + num_base_cols + score_cols

X_train_with_scores = train_df[features_with_scores]
X_test_with_scores = test_df[features_with_scores]
```

### Pipeline para codificacion de variables categoricas e imputacion de nulos

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from xgboost import XGBClassifier

def build_pipeline(num_cols, cat_cols, classifier):

    num_transformer = Pipeline(steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ])

    cat_transformer = Pipeline(steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ])

    preprocessor = ColumnTransformer(
        transformers=[
            ("num", num_transformer, num_cols),
            ("cat", cat_transformer, cat_cols)
        ]
    )
    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", classifier)
        ])

    return pipeline
```

### Primer modelo XGBClassifier con GridSearchCV.
No se excluyen los scores de las categorias. Potencial **leakage**

```python
#Modelo con scores con busqueda de hiperparametros
from sklearn.model_selection import GridSearchCV
num_cols_with_scores = num_base_cols + score_cols

model_xgb_with_scores = build_pipeline(
    num_cols=num_cols_with_scores,
    cat_cols=cat_cols,
    classifier=XGBClassifier(
        eval_metric="logloss",
        random_state=42
    )
)

param_grid_xgb = {
    "classifier__n_estimators": [150, 200, 250],
    "classifier__max_depth": [3, 5],
    "classifier__learning_rate": [0.05, 0.1, 0.15],
    "classifier__subsample": [0.8]
}

grid_xgb_with_scores = GridSearchCV(
    estimator=model_xgb_with_scores,
    param_grid=param_grid_xgb,
    scoring="roc_auc",
    cv=5,
    n_jobs=-1,
    verbose=1
)

grid_xgb_with_scores.fit(X_train_with_scores, y_train)
```

## **Selección y Justificación de Modelos**

Se evaluaron distintas alternativas de clasificación para predecir la satisfacción de los pasajeros a partir de variables demográficas, operativas y de contexto del viaje. Frente a modelos lineales tradicionales, como la Regresión Logística, y modelos basados en proximidad, como K-Nearest Neighbors (K-NN), se priorizó el uso de métodos de ensamble basados en árboles de decisión debido a su capacidad para capturar relaciones no lineales e interacciones complejas entre variables sin requerir supuestos estrictos sobre la distribución de los datos.

Se seleccionaron Random Forest y XGBoost, dos arquitecturas de ensamble basadas en árboles CART (Classification and Regression Trees). Ambos algoritmos permiten modelar relaciones complejas entre variables categóricas, ordinales y numéricas, manteniendo un adecuado equilibrio entre capacidad predictiva y robustez frente a datos heterogéneos.

Random Forest implementa una estrategia de bagging, entrenando múltiples árboles independientes sobre subconjuntos aleatorios de observaciones y atributos para reducir la varianza del estimador. XGBoost, en cambio, utiliza una estrategia de gradient boosting, construyendo secuencialmente árboles que corrigen los errores cometidos por los anteriores mediante la optimización iterativa de una función de pérdida.

Si bien ambos modelos presentan un menor nivel de interpretabilidad respecto de modelos lineales, su desempeño predictivo puede analizarse mediante técnicas de importancia de variables, permitiendo comprender qué atributos influyen con mayor intensidad en las decisiones del modelo.

Durante la etapa de preprocesamiento se decidió no aplicar técnicas de escalado o normalización sobre las variables numéricas. Esta decisión se fundamenta en que los modelos basados en árboles realizan sus particiones mediante reglas de decisión basadas en umbrales y no mediante métricas de distancia. En consecuencia, variables con escalas muy diferentes, como Age y Flight Distance, pueden coexistir sin afectar significativamente el desempeño del algoritmo. Esta característica permitió simplificar el flujo de preparación de datos sin esperar un impacto relevante sobre la capacidad predictiva de los modelos evaluados.

## **Tratamiento de Variables de Percepción del Servicio**

Durante una primera etapa de experimentación se entrenó un modelo XGBoost incorporando variables de evaluación del servicio reportadas por los pasajeros, tales como Online Boarding, Inflight WiFi Service, Seat Comfort, Food and Drink e Inflight Entertainment, entre otras.

Inicialmente, este enfoque produjo métricas predictivas excepcionalmente elevadas. Sin embargo, un análisis posterior permitió identificar que dichas variables provenían de la misma encuesta utilizada para construir la variable objetivo (satisfaction). En otras palabras, el modelo estaba utilizando evaluaciones parciales del servicio para predecir la evaluación global del mismo pasajero.

Esta situación introduce un fenómeno de leakage conceptual, ya que el modelo aprende a partir de atributos estrechamente relacionados con el mismo proceso de evaluación que genera la variable objetivo.

Debido a ello, se decidió descartar estas variables del análisis definitivo y reconstruir el modelo XGBoost utilizando exclusivamente variables operativas, demográficas y de contexto del viaje. Esta segunda versión representa de forma más fiel un escenario real de predicción de satisfacción.

Siguiendo el mismo criterio metodológico, el modelo Random Forest fue entrenado únicamente utilizando el conjunto depurado de variables, sin incorporar las variables de evaluación del servicio.

De esta manera, la comparación final entre algoritmos se realizó sobre un mismo conjunto de características libres de leakage conceptual, garantizando una evaluación más rigurosa y representativa de la capacidad de generalización de los modelos.

```python
import mlflow
import mlflow.sklearn
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from mlflow import MlflowClient
from mlflow.entities import ViewType
from mlflow.models import infer_signature

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    RocCurveDisplay,
)

TRACKING_URI = "http://localhost:5001"
EXPERIMENT_NAME = "tp_airline_satisfaction"

REGISTRY_WITH_SCORES = "airline_satisfaction_with_scores"
REGISTRY_WITHOUT_SCORES = "airline_satisfaction_without_scores"

# Evita el error de intentar abrir un run cuando quedó otro activo.
if mlflow.active_run() is not None:
    mlflow.end_run()

mlflow.set_tracking_uri(TRACKING_URI)
client = MlflowClient(tracking_uri=TRACKING_URI)

print("MLflow:", mlflow.__version__)
print("Tracking URI:", mlflow.get_tracking_uri())

# Esta llamada también valida que el servidor esté disponible.
client.search_experiments(max_results=1)
print("Conexión con MLflow correcta.")
```

```python
def get_or_create_active_experiment(experiment_name: str) -> str:
    """Crea, recupera o restaura un experimento borrado lógicamente."""

    experiments = client.search_experiments(
        view_type=ViewType.ALL,
        filter_string=f"name = '{experiment_name}'",
        max_results=10,
    )

    if not experiments:
        experiment_id = client.create_experiment(experiment_name)
        print(f"Experimento creado: {experiment_name} (ID={experiment_id})")
        return experiment_id

    experiment = experiments[0]

    if experiment.lifecycle_stage == "deleted":
        client.restore_experiment(experiment.experiment_id)
        print(
            f"Experimento restaurado: {experiment_name} "
            f"(ID={experiment.experiment_id})"
        )
    else:
        print(
            f"Experimento existente: {experiment_name} "
            f"(ID={experiment.experiment_id})"
        )

    return experiment.experiment_id


EXPERIMENT_ID = get_or_create_active_experiment(EXPERIMENT_NAME)
mlflow.set_experiment(experiment_id=EXPERIMENT_ID)
```

```python
def calculate_binary_metrics(model, X, y_true):
    y_pred = model.predict(X)
    y_proba = model.predict_proba(X)[:, 1]

    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

    metrics = {
        "accuracy": float(accuracy_score(y_true, y_pred)),
        "precision": float(precision_score(y_true, y_pred, zero_division=0)),
        "recall": float(recall_score(y_true, y_pred, zero_division=0)),
        "f1_score": float(f1_score(y_true, y_pred, zero_division=0)),
        "roc_auc": float(roc_auc_score(y_true, y_proba)),
        "false_positive_rate": float(fp / (fp + tn)) if (fp + tn) else 0.0,
        "false_negative_rate": float(fn / (fn + tp)) if (fn + tp) else 0.0,
    }

    return metrics, y_pred, y_proba


def create_confusion_matrix_figure(y_true, y_pred, title):
    fig, ax = plt.subplots(figsize=(7, 5))
    ConfusionMatrixDisplay.from_predictions(
        y_true,
        y_pred,
        display_labels=["Neutral/insatisfecho", "Satisfecho"],
        ax=ax,
    )
    ax.set_title(title)
    fig.tight_layout()
    return fig


def create_roc_figure(y_true, y_proba, title):
    fig, ax = plt.subplots(figsize=(7, 5))
    RocCurveDisplay.from_predictions(y_true, y_proba, ax=ax)
    ax.plot([0, 1], [0, 1], linestyle="--")
    ax.set_title(title)
    ax.grid(True)
    fig.tight_layout()
    return fig


def create_feature_importance_figure(model, top_n=15):
    classifier = model.named_steps["classifier"]
    preprocessor = model.named_steps["preprocessor"]

    if not hasattr(classifier, "feature_importances_"):
        return None, None

    feature_names = preprocessor.get_feature_names_out()
    importance_df = (
        pd.DataFrame(
            {
                "feature": feature_names,
                "importance": classifier.feature_importances_,
            }
        )
        .sort_values("importance", ascending=False)
        .reset_index(drop=True)
    )

    top_df = importance_df.head(top_n).sort_values("importance")

    fig, ax = plt.subplots(figsize=(9, 6))
    ax.barh(top_df["feature"], top_df["importance"])
    ax.set_title(f"Top {top_n} variables más importantes")
    ax.set_xlabel("Importancia")
    fig.tight_layout()

    return fig, importance_df
```

```python
def log_and_register_model(
    *,
    model,
    X_test,
    y_test,
    grid_search,
    run_name,
    registry_name,
    model_family,
    feature_set,
    extra_tags=None,
):
    """Registra métricas, artefactos, pipeline y una versión en Model Registry."""

    tags = {
        "project": "TP aprendizaje de máquina - Grupo 8",
        "task": "binary_classification",
        "target": "satisfaction",
        "model_family": model_family,
        "feature_set": feature_set,
    }
    tags.update(extra_tags or {})

    metrics, y_pred, y_proba = calculate_binary_metrics(model, X_test, y_test)
    metrics["cv_best_roc_auc"] = float(grid_search.best_score_)

    params = {
        key: value
        for key, value in grid_search.best_params_.items()
    }

    input_example = X_test.head(5).copy()
    signature = infer_signature(
        X_test.head(50),
        model.predict(X_test.head(50)),
    )

    with mlflow.start_run(
        experiment_id=EXPERIMENT_ID,
        run_name=run_name,
        tags=tags,
    ) as run:
        mlflow.log_params(params)
        mlflow.log_metrics(metrics)

        cm_fig = create_confusion_matrix_figure(
            y_test,
            y_pred,
            f"Matriz de confusión - {run_name}",
        )
        mlflow.log_figure(cm_fig, "plots/confusion_matrix.png")
        plt.close(cm_fig)

        roc_fig = create_roc_figure(
            y_test,
            y_proba,
            f"Curva ROC - {run_name}",
        )
        mlflow.log_figure(roc_fig, "plots/roc_curve.png")
        plt.close(roc_fig)

        importance_fig, importance_df = create_feature_importance_figure(model)

        if importance_fig is not None:
            mlflow.log_figure(
                importance_fig,
                "plots/feature_importance.png",
            )
            plt.close(importance_fig)

            mlflow.log_table(
                importance_df,
                artifact_file="tables/feature_importance.json",
            )

        model_info = mlflow.sklearn.log_model(
            sk_model=model,
            name="model",
            registered_model_name=registry_name,
            signature=signature,
            input_example=input_example,
            serialization_format="cloudpickle",
            metadata={
                "target": "satisfaction",
                "feature_set": feature_set,
            },
        )

        version = getattr(model_info, "registered_model_version", None)

        if version is None:
            versions = client.search_model_versions(
                filter_string=f"name = '{registry_name}'"
            )
            matching = [
                item for item in versions
                if item.run_id == run.info.run_id
            ]
            if not matching:
                raise RuntimeError(
                    f"No se encontró la versión registrada para el run {run.info.run_id}"
                )
            version = matching[0].version

        client.set_model_version_tag(
            name=registry_name,
            version=str(version),
            key="feature_set",
            value=feature_set,
        )
        client.set_model_version_tag(
            name=registry_name,
            version=str(version),
            key="roc_auc",
            value=f"{metrics['roc_auc']:.6f}",
        )

        return {
            "run_id": run.info.run_id,
            "registry_name": registry_name,
            "version": str(version),
            "roc_auc": metrics["roc_auc"],
            "accuracy": metrics["accuracy"],
            "model_family": model_family,
            "feature_set": feature_set,
        }
```

```python
# Descripciones y tags de los modelos registrados.
client.update_registered_model(
    name=REGISTRY_WITHOUT_SCORES,
    description=(
        "Modelos candidatos para predecir satisfacción sin utilizar los scores "
        "de servicio. XGBoost y Random Forest comparten el mismo esquema de entrada."
    ),
)

client.set_registered_model_tag(
    REGISTRY_WITHOUT_SCORES,
    "project",
    "TP aprendizaje de máquina - Grupo 8",
)
client.set_registered_model_tag(
    REGISTRY_WITHOUT_SCORES,
    "feature_set",
    "without_scores",
)

client.update_registered_model(
    name=REGISTRY_WITH_SCORES,
    description=(
        "Modelo XGBoost que utiliza scores de servicio. Se registra como referencia "
        "separada porque tiene otro esquema de entrada y posible data leakage."
    ),
)

client.set_registered_model_tag(
    REGISTRY_WITH_SCORES,
    "project",
    "TP aprendizaje de máquina - Grupo 8",
)
client.set_registered_model_tag(
    REGISTRY_WITH_SCORES,
    "potential_data_leakage",
    "true",
)

# Champion/challenger solamente entre modelos comparables: mismo dataset de entrada.
without_scores_results = (
    registration_df[
        registration_df["registry_name"] == REGISTRY_WITHOUT_SCORES
    ]
    .sort_values("roc_auc", ascending=False)
    .reset_index(drop=True)
)

champion = without_scores_results.iloc[0]
challenger = without_scores_results.iloc[1]

client.set_registered_model_alias(
    REGISTRY_WITHOUT_SCORES,
    "champion",
    champion["version"],
)
client.set_registered_model_alias(
    REGISTRY_WITHOUT_SCORES,
    "challenger",
    challenger["version"],
)

with_scores_result = registration_df[
    registration_df["registry_name"] == REGISTRY_WITH_SCORES
].iloc[0]

client.set_registered_model_alias(
    REGISTRY_WITH_SCORES,
    "reference",
    with_scores_result["version"],
)

print(
    "Champion:",
    champion["model_family"],
    "versión",
    champion["version"],
    "ROC AUC",
    round(champion["roc_auc"], 4),
)
print(
    "Challenger:",
    challenger["model_family"],
    "versión",
    challenger["version"],
    "ROC AUC",
    round(challenger["roc_auc"], 4),
)
print(
    "Modelo con scores registrado como referencia, versión",
    with_scores_result["version"],
)
```
